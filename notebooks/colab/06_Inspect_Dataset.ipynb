# 06 Inspect Dataset

## Objective
Inspect Healthy_001 with the existing read-only Dataset Adapter.

## Prerequisites
Run notebook 05 or place an approved manifest-backed dataset at the documented path.

## Expected Output
Manifest, metadata, file, checksum, trajectory, and frame-count checks.

## Troubleshooting
WAITING_DATASET means no manifest was found; INVALID findings identify the exact integrity issue.

## Validation
The inspection cell reports schema, checksum, completeness, and finite-value checks.

```python
import os
from pathlib import Path
repo = Path.cwd() / 'drosophila-pd-flygym'
if (repo / 'pyproject.toml').is_file():
    os.chdir(repo)
try:
    from drosophila_pd.dataset_adapter import DatasetValidator, FlyGymDataset
    manifest = Path('datasets/healthy/Healthy_001/manifest.json')
    if manifest.is_file():
        dataset = FlyGymDataset.from_manifest(manifest)
        report = DatasetValidator().validate(dataset)
        print(report.as_dict())
    else:
        print('WAITING_DATASET: Healthy_001 manifest is not present.')
except Exception as exc:
    print('Dataset inspection failed:', type(exc).__name__, exc)
```
