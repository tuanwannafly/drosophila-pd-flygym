# 02 Create First Fly

## Objective
Create the first FlyGym fly through FlyGymAdapter and inspect metadata.

## Prerequisites
FlyGym 2.1.0 and MuJoCo 3.9.0 must be installed.

## Expected Output
A NeuroMechFly-compatible object, its name, class, and pre-materialization state.

## Troubleshooting
If the canonical helper is unavailable, rerun notebook 00 with the pinned requirements.

## Validation
The cell prints the constructed FlyGym object and its metadata before continuing.

## Next notebook

```python
import os
from pathlib import Path
repo = Path.cwd() / 'drosophila-pd-flygym'
if (repo / 'pyproject.toml').is_file():
    os.chdir(repo)
try:
    from drosophila_pd.flygym_adapter import FlyGymAdapter, FlyGymConfig
    config = FlyGymConfig.from_yaml('configs/v2/flygym/healthy.yaml')
    fly = FlyGymAdapter().create_fly(config.fly)
    print('class:', type(fly).__module__ + '.' + type(fly).__name__)
    print('name:', fly.name)
    print('body segments:', len(fly.get_bodysegs_order()))
    print('skeleton before materialization:', fly.skeleton)
except Exception as exc:
    print('Fly creation failed:', type(exc).__name__, exc)
```
