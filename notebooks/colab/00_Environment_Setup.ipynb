# 00 Environment Setup

## Objective
Prepare a clean Google Colab runtime for the repository.

## Prerequisites
A fresh Google Colab Python runtime with internet access.

## Expected Output
The repository, Python package, FlyGym 2.1.0, and MuJoCo 3.9.0 are checked.

## Troubleshooting
If installation fails, read the printed command and restart the runtime before retrying.

## Validation
The final lines report package versions and import status; a missing package is a setup failure.

## Next notebook
01_FlyGym_API_Explorer.ipynb

```python
import importlib.metadata as metadata
import importlib.util
import os
from pathlib import Path
import subprocess
import sys

repo = Path.cwd() / 'drosophila-pd-flygym'
def attempt(command):
    print('RUN:', ' '.join(str(item) for item in command))
    result = subprocess.run(command, check=False)
    print('returncode:', result.returncode)
    return result.returncode == 0

if not (repo / 'pyproject.toml').is_file():
    attempt(['git', 'clone', 'https://github.com/TanVi3001/drosophila-pd-flygym.git', str(repo.name)])
if (repo / 'pyproject.toml').is_file():
    os.chdir(repo)
    attempt([sys.executable, '-m', 'pip', 'install', '-e', '.[test]'])
    attempt([sys.executable, '-m', 'pip', 'install', '-r', 'requirements-colab.txt'])
else:
    print('Repository checkout is unavailable; fix the clone step and rerun this notebook.')

def show_version(name):
    try:
        print(name, metadata.version(name))
    except metadata.PackageNotFoundError as exc:
        print(name, 'NOT INSTALLED:', exc)

show_version('flygym')
show_version('mujoco')
print('drosophila_pd importable:', importlib.util.find_spec('drosophila_pd') is not None)
try:
    import flygym
    import mujoco
    import drosophila_pd
    print('Imports verified:', flygym.__name__, mujoco.__name__, drosophila_pd.__name__)
except Exception as exc:
    print('Import verification failed:', type(exc).__name__, exc)
```
